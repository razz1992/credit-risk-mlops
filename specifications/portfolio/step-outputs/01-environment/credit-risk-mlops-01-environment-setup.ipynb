## 3. Acquire the UCI dataset

```python
from ucimlrepo import fetch_ucirepo

# Fetch dataset 350 from the UCI Machine Learning Repository
dataset = fetch_ucirepo(id=350)

# Separate independent variables and target
X = dataset.data.features
y = dataset.data.targets

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
df = pd.concat([X, y], axis=1)

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Complete dataset shape:", df.shape)
display(df.head())
```

## 4. Inspect the schema

## 5. Check missing values and duplicate rows

```python
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])

print("Duplicate rows:", df.duplicated().sum())
```

## 6. Inspect target distribution

```python
target_column = y.columns[0]

print("Target columns:", y.columns.tolist())
print("\nTarget distribution:")
print(y[target_column].value_counts())
print("\nTarget percentage:")
print(y[target_column].value_counts(normalize=True) * 100)
```
